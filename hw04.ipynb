Flood risk is usually quantified by propagating a distribution of flood
depths through a depth-damage function, which converts a dflood epth
into an economic loss. A common form for a house without a basement is a
bounded logistic curve,

$$d(h) = \mathbb{1}_{h > 0} \frac{L}{1 + \exp\left(-k(h - h_0)\right)},$$

where $h$ is the flood depth in metres and $d$ is the damage in dollars.
We’ll use $L = \$200,000$, $k = 0.8$, and $h_0 = 3$ in our depth-damage
function specification.

Assume the annual maximum flood depth at this structure is distributed
as

$$h \sim \text{LogNormal}(1.2,\ 0.3).$$

#### Problem 2.1 (5 points)

Implement the depth-damage function and plot it over depths from 0 to
8m. On a second panel, plot the density of the hazard distribution over

Looking at the two panels together: over what range of depths does a

#### Problem 2.2 (8 points)

Estimate the expected annual damage using Monte Carlo.

Report your estimate, a 95% confidence interval, the sample size you
used, and the seed you set. Then justify the sample size by doubling it
and comparing the estimates.

```julia

using Random
using Distributions
using Statistics
using Plots

## Implement the depth-damage function and plot it over depths from 0 to 8m
# Given parameters 
flood_params = let 
    L = 200000
    k = 0.8 
    h0 = 3 
    (; L, k, h0)
end

# Given the log normal distribution of h 
distribution = LogNormal(1.2, 0.3)


function depth_damage(depth; params = flood_params) # input any depth 
    
    # A depth at or below zero causes no damage.
    if depth <= 0
        return 0
    end
    # TODO: return the damage in dollars for a flood of this depth.
    return 1* (params.L / (1 + exp(- params.k * (depth - params.h0)))) #d(h) function given 
    #error("depth_damage has not been written yet -- replace this line")
end

depths_asked = range(0, 8, length = 1000) # Need to set length = N to get smooth plot
result_damage = [depth_damage(h) for h in depths_asked]


### Plot 

damage_plot = plot(depths_asked, result_damage, xlabel="Flood depth (m)", ylabel="Damage (dollars)", legend=false)


## On a second panel, plot the density of the hazard distribution over the same range.
### Plot 
density_hazard_plot = plot(depths_asked, pdf.(distribution, depths_asked), xlabel="Flood depth (m)", ylabel="Density of hazard", legend=false)


## Show both panels
plot(damage_plot, density_hazard_plot, layout = (2, 1))
```

### Answer to Problem 2.1 

The largest chnage in damage over a small change in depth is reflected in the slope of the curve on flood depth vs damage plot. The range of large slope appears to happen between flood depth of 1 to 5m from the depth vs damage plot. To narrow down the range, we can use the second plot of density of hazard to see that the there is a greatest density of hazard around 3m of flood depth. This means that the steepest pat of the damage curve is around depth h = 3m. Therefore, **the range of about h = 2.5 to 3.5m is where a small change in depth produce the largest change in damange.**  


#### Problem 2.2 (8 points)

Estimate the expected annual damage using Monte Carlo.

Report your estimate, a 95% confidence interval, the sample size you
used, and the seed you set. Then justify the sample size by doubling it
and comparing the estimates.

```julia
using Random # Need to run this every time to get same seed
using Distributions
using Statistics
using Plots

### Use Monte Carlo 

seed  = 4750 # Same as Prof's choice and just because :)  
number_of_samples = 100000  # Large sample size to be accurate, start with 10,000 because after testing a few times of doubling sample size, this is the value when CI stay relatively similiar

# Given the log normal distribution of h 
# The uncertain input 
distribution = LogNormal(1.2, 0.3)


function damage_estimate(number_of_samples, seed)
    Random.seed!(seed)
    sample_depths = rand(distribution, number_of_samples)
    damage = zeros(number_of_samples)
    for i in 1:number_of_samples
        damage[i] = depth_damage(sample_depths[i])
    end
    return damage
end

results = damage_estimate(number_of_samples, seed)


### Results to report 

estimate = mean(results) # estimate
println(estimate) 

# Need SE to get CI 
se = std(results)/sqrt(number_of_samples) # Using the formula for SE as the standard devision over the sqrt of sample size
ci_low= (estimate - 1.96*se) # 95% Confidence interval lower bound 
ci_up= (estimate + 1.96*se) # 95% Confidence interval upper bound 

ci = (ci_low, ci_up) 
println(ci)

#################################### Code to answer P2.3 #################################### 

println()

p99_damage = quantile(results, 0.99)
println(p99_damage)
diff_from_mean = p99_damage - estimate
println(diff_from_mean)
ratio_to_mean = p99_damage / estimate
println(ratio_to_mean)
```

```julia
######## Run again with the doubled sample size ########

seed  = 4750 # Same as Prof's choice and just because :)  
number_of_samples = 20000  # Doubled sample size 

# Given the log normal distribution of h 
# The uncertain input 
distribution = LogNormal(1.2, 0.3)


function damage_estimate(number_of_samples, seed)
    Random.seed!(seed)
    sample_depths = rand(distribution, number_of_samples)
    damage = zeros(number_of_samples)
    for i in 1:number_of_samples
        damage[i] = depth_damage(sample_depths[i])
    end
    return damage
end

results = damage_estimate(number_of_samples, seed)


### Results to report 

estimate = mean(results) # estimate
println(estimate) 

# Need SE to get CI 
se = std(results)/sqrt(number_of_samples) # Using the formula for SE as the standard devision over the sqrt of sample size
ci_low= (estimate - 1.96*se) # 95% Confidence interval lower bound 
ci_up= (estimate + 1.96*se) # 95% Confidence interval upper bound 

ci = (ci_low, ci_up) 
println(ci)
```
